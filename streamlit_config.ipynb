```python
import os

os.makedirs(".streamlit", exist_ok=True)
```

```python
config = """
[server]
headless = true
address = "0.0.0.0"
port = 7860
enableXsrfProtection = false
enableCORS = false

[browser]
gatherUsageStats = false
"""

with open(".streamlit/config.toml", "w") as f:
    f.write(config)

print("config.toml created successfully")
```
